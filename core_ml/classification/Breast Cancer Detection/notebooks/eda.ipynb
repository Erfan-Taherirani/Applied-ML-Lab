```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Load the dataset
data = load_breast_cancer()

# Create a DataFrame
df_full = pd.DataFrame(data.data, columns=data.feature_names)

# Add the target variable to the DataFrame
df_full['target'] = data.target

# drop the highly correlated features
df_full = df_full.drop(
    labels=[
        "mean perimeter", "mean area", "perimeter error",
        "area error", "worst perimeter", "worst area"
	],
    axis=1
)

# drop mean fractal dimension (very low correlation)
df_full = df_full.drop(
    labels=[
        "mean fractal dimension"
	],
    axis=1
)
df_full
```

there are tow types of breast cancer
1. benign
2. malignant

the target variable is binary:
1. malignant = 0
2. benign = 1

```python
df = df_full.filter(regex="mean|target")
```

```python
g = sns.displot(
    data=df,
    x="mean radius",
    hue="target"
)
g.figure.axes[0].set_title("Distribution of Mean Radius by Target Class")
```
