```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv("../data/glass.csv")

# fixing data types
df['Type'] = df['Type'].astype('category')

# removing K and Ca columns (very low correlation)
df = df.drop(["K", "Ca"], axis=1)
```

```python
sns.catplot(
    data=df,
    x="Type",
    kind="count",
)
```

```python
print("The Percentage of each type of glass is as follows")
df['Type'].value_counts().sort_values(
    ascending=False
) / df.shape[0]
```
