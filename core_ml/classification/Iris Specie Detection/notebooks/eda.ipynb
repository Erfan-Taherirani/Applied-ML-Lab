# Data Exploration:

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = sns.load_dataset("iris")
df['species'] = df['species'].astype("category")

df['species'] = df['species'].map({
    "setosa": 0,
    "versicolor": 1,
    "virginica": 2
})
```

```python
np.random.seed(42)

choices = ["petal_length", "petal_width"]
label_to_drop = np.random.choice(choices)
df = df.drop(str(label_to_drop), axis=1)

df
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
g = sns.displot(
    data=df,
    x="sepal_length",
    hue="species",
    bins=15
)
g.ax.set(
    title="Sepal Length Distribution by Species",
)
```

```python
g = sns.displot(
    data=df,
    x="sepal_width",
    hue="species",
    fill=True,
    kind="kde"
)
g.ax.set(
    title="Sepal Width Distribution by Species"
)
```

```python
g = sns.displot(
    data=df,
    x="petal_width",
    hue="species",
    bins=30
)
g.ax.set(
    title="Petal Width Distribution by Species"
)
```

```python
g = sns.heatmap(
    data=abs(df.corr()),
    annot=True,
    cmap="Reds"
)
g.axes.set(
    title="Correlation Matrix (absolute values)"
)
```

```python
g = sns.heatmap(
    data=df.corr(),
    annot=True,
    cmap="coolwarm"
)
g.axes.set(
    title="Correlation Matrix (Original values)"
)
```
